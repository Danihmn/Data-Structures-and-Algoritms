```python
def frequence_in(s: str):
    freq = {}
    for ch in s:
        if ch in freq:
            freq[ch] += 1
        else:
            freq[ch] = 1
    return freq
```

```python
name = "Ana Clara Frasom Torquato Bezerra"
frequence_in_name = frequence_in(name)
```
